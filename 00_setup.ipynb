# LH Nautical - Desafio Lighthouse de Dados e IA

## 00 - Configuração do Projeto

Este notebook prepara o ambiente e organiza os dados utilizados no desafio.

### Como os dados serão organizados

Vamos trabalhar em três etapas:

- **Bronze (o depósito):** guarda os arquivos como foram recebidos, sem perder a fonte original.
- **Silver (a área de organização)** limpa, corrige e padroniza os dados.
- **Gold (a vitrine)** reúne os dados já preparados para análises, indicadores e modelos.

Assim, conseguimos sempre voltar à origem caso alguma regra precise ser revisada.

### Dados disponíveis

O desafio possui 24 arquivos CSV com informações da LH Nautical entre 2020 e 2026, incluindo clientes, produtos, vendas, pagamentos, compras, estoque e devoluções.

### Primeiro passo

Os arquivos originais serão armazenados antes de qualquer transformação. Depois, avançaremos gradualmente da organização dos dados até as análises de negócio, previsão de demanda e recomendações.
